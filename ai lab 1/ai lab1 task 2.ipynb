```python
# Task 1 - Question 2: rule based screening
# Score 70 and attendance 75% are assumed values for testing only

def rule_based(score, documents_complete, attendance, discipline_clear):
    reasons = []
    if score < 70:
        reasons.append("score is less than 70")
    if not documents_complete:
        reasons.append("documents are not complete")
    if attendance < 75:
        reasons.append("attendance is less than 75%")
    if not discipline_clear:
        reasons.append("discipline record is not clear")

    if reasons:
        return "Hold", ", ".join(reasons)
    return "Review", "all conditions are pass"


# (case, score, documents_complete, attendance, discipline_clear, expected)
cases = [
    ("Case1", 85, True, 80, True,  "Review"),
    ("Case2", 85, True, 70, True,  "Hold"),
    ("Case3", 85, True, 80, False, "Hold"),
    ("Case4", 70, True, 75, True,  "Review"),
]

for name, score, docs, att, disc, expected in cases:
    actual, reason = rule_based(score, docs, att, disc)
    print(name)
    print("Inputs   : score", score, ", documents complete", docs,
          ", attendance", att, ", discipline clear", disc)
    print("Expected :", expected)
    print("Actual   :", actual, "(matched)" if actual == expected else "(NOT matched)")
    print("Reason   :", reason)
    print()
```
