```python
from collections import deque

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["G"],
    "F": [],
    "G": []
}

def bfs(start, goal):
    q = deque([(start, [start])])
    seen = {start}

    while q:
        node, path = q.popleft()

        if node == goal:
            return path

        for n in graph[node]:
            if n not in seen:
                seen.add(n)
                q.append((n, path + [n]))

def dfs(start, goal):
    stack = [(start, [start])]
    seen = set()

    while stack:
        node, path = stack.pop()

        if node == goal:
            return path

        if node in seen:
            continue

        seen.add(node)

        for n in reversed(graph[node]):
            stack.append((n, path + [n]))

print("BFS:", bfs("A", "G"))
print("DFS:", dfs("A", "G"))
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["G"],
    "F": [],
    "G": []
}

def dls(node, goal, limit, path):
    if node == goal:
        return "FOUND", path

    if len(path) - 1 == limit:
        for child in graph[node]:
            if child not in path:
                return "CUTOFF", None
        return "FAILURE", None

    cutoff = False

    for child in graph[node]:
        if child not in path:
            status, result = dls(
                child, goal, limit, path + [child]
            )

            if status == "FOUND":
                return status, result

            if status == "CUTOFF":
                cutoff = True

    if cutoff:
        return "CUTOFF", None

    return "FAILURE", None


def iddfs(start, goal, max_depth):
    for limit in range(max_depth + 1):
        status, path = dls(start, goal, limit, [start])

        if status == "FOUND":
            return status, path, limit

    return "CUTOFF", None, max_depth


print("DLS RESULTS")

for limit in range(1, 5):
    status, path = dls("A", "G", limit, ["A"])
    print("Limit:", limit, "Status:", status, "Path:", path)


print("\nIDDFS RESULTS")

for maximum in range(1, 5):
    status, path, depth = iddfs("A", "G", maximum)
    print("Maximum:", maximum, "Status:", status,
          "Path:", path)
```
